## Cria tabele Setor

```python
avarias = (
    silverPerdas
    .groupby('SETOR_ORIGEM', as_index=False)['QTD_AVARIADA']
    .sum()
)
```

```python
prejuizo = (
    silverPerdas
    .groupby('SETOR_ORIGEM', as_index=False)['V_PREJUIZO']
    .sum()
)
```

```python
g_perdas_setor = pd.concat(
    [avarias['SETOR_ORIGEM'],avarias['QTD_AVARIADA'], prejuizo['V_PREJUIZO']],
    axis=1
)
```

```python
g_perdas_setor.head(5)
```
