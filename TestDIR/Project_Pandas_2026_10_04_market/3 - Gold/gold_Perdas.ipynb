## Cria tabele Setor

```python
g_avarias = silverPerdas.groupby('SETOR_ORIGEM', as_index=False).agg(
    QTD_AVARIADA = ('QTD_AVARIADA', 'sum'),
    V_PREJUIZO = ('V_PREJUIZO', 'sum'),
)
```

```python
g_avarias.head(5)
```

## Cria tabele Responsável

```python
g_Responsavel_prejuizo = silverPerdas.groupby('RESPONSAVEL', as_index=False).agg(
    QTD_AVARIADA = ('QTD_AVARIADA', 'sum'),
    V_PREJUIZO = ('V_PREJUIZO', 'sum'),
)
```

```python
g_Responsavel_prejuizo.head(3)
```

-----

## Cria tabele ANO

```python
g_ano_prejuizo = silverPerdas.groupby('ANO', as_index=False).agg(
    QTD_AVARIADA = ('QTD_AVARIADA', 'sum'),
    V_PREJUIZO = ('V_PREJUIZO', 'sum'),
)
```

```python
g_ano_prejuizo.head(5)
```

-------------

## Cria tabele MES

```python
g_mes_prejuizo = silverPerdas.groupby('MES', as_index=False).agg(
    QTD_AVARIADA = ('QTD_AVARIADA', 'sum'),
    V_PREJUIZO = ('V_PREJUIZO', 'sum'),
)
```

```python
g_mes_prejuizo.head(5)
```

------

## Cria tabele COD_PROD

```python
g_cod_prejuizo = silverPerdas.groupby('COD_PROD', as_index=False).agg(
    QTD_AVARIADA = ('QTD_AVARIADA', 'sum'),
    V_PREJUIZO = ('V_PREJUIZO', 'sum'),
)
```

```python
gold_Perdas = silverPerdas

def export_dados_excel(resultado):

    pasta = Path("..\\Raw\\table_gold\\")
    pasta.mkdir(parents=True, exist_ok=True)

    if resultado:

        arquivo_excel = pasta / "gold_Perdas.xlsx"

        # 1. Criar os arquivos
        gold_Perdas.to_excel(arquivo_excel, index=False)


        # 3. Consolidar em um Excel com duas abas
        with pd.ExcelWriter(arquivo_excel, engine="openpyxl") as writer:

            gold_Perdas.to_excel(
                writer,
                sheet_name="g_Perdas",
                index=False
            )


            g_avarias.to_excel(
                writer,
                sheet_name="avarias",
                index=False
            )

            g_Responsavel_prejuizo.to_excel(
                writer,
                sheet_name="resposavel",
                index=False
            )

            g_mes_prejuizo.to_excel(
                writer,
                sheet_name="mes",
                index=False
            )

            g_ano_prejuizo.to_excel(
                writer,
                sheet_name="ano",
                index=False
            )

            g_cod_prejuizo.to_excel(
                writer,
                sheet_name="cod_prod",
                index=False
            )

            
        return True

    print("ERRO")
    return False
```

```python
export_dados_excel(True)
```
