# Cria tabela de agrupamento por coluna

### cria fornecedor

```python
gold_fornecedor_qtd = (
    GoldComp
    .groupby('FORNECEDOR', as_index=False)['QTD_COMPRADA']
    .sum()
)
gold_fornecedor_qtd.head(5)
```

```python
GoldComp['TOTAL_NOTA'] = pd.to_numeric(GoldComp['TOTAL_NOTA'].astype(str).str.replace('R$ ',''))
```

```python
gold_fornecedor_valor = (
    GoldComp
    .groupby('FORNECEDOR', as_index=False)['TOTAL_NOTA']
    .sum()
)
```

```python
gold_fornecedor_valor.head(5)
```

```python
gold_fornecedor = pd.concat(
    [gold_fornecedor_qtd['FORNECEDOR'],gold_fornecedor_qtd['QTD_COMPRADA'],gold_fornecedor_valor['TOTAL_NOTA']],
    axis=1
)
gold_fornecedor.head(3)
```
