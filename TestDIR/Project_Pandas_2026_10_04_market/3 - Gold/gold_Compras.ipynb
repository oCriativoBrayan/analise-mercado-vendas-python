# Cria tabela de agrupamento por coluna

### cria fornecedor

```python
gold_fornecedor = silverCompras.groupby('FORNECEDOR', as_index=False).agg(
      QTD_COMPRADA=('QTD_COMPRADA', 'sum'),
      TOTAL_NOTA=('TOTAL_NOTA', 'sum'),
  )
gold_fornecedor.head(5)
```

## Cria Cod_Prod

```python
gold_cod_qtd = silverCompras.groupby('COD_PROD', as_index=False).agg(
    QTD_COMPRADA=('QTD_COMPRADA', 'sum'),
    TOTAL_NOTA=('TOTAL_NOTA', 'sum'),
)
gold_cod_qtd.head(5)
```

### Cria Responsável

```python
gold_resp = silverCompras.groupby('RESPONSAVEL_RECEBIMENTO', as_index=False).agg(
    QTD_COMPRADA = ('QTD_COMPRADA', 'sum'),
    TOTAL_NOTA = ('TOTAL_NOTA','sum'),
)
gold_resp.head(5)
```

### Cria mes

```python
gold_mes = silverCompras.groupby('MES', as_index=False).agg(
    QTD_COMPRADA = ('QTD_COMPRADA', 'sum'),
    TOTAL_NOTA = ('TOTAL_NOTA','sum'),
)
gold_mes.head(5)
```

### Cria por ano

```python
gold_ano = silverCompras.groupby('ANO', as_index=False).agg(
    QTD_COMPRADA = ('QTD_COMPRADA', 'sum'),
    TOTAL_NOTA = ('TOTAL_NOTA','sum'),
)
gold_ano.head(5)
```

------

## Pela N_NF

```python
gold_nf = silverCompras.groupby('N_NF', as_index=False).agg(
    QTD_COMPRADA = ('QTD_COMPRADA', 'sum'),
    TOTAL_NOTA = ('TOTAL_NOTA','sum'),
)
gold_nf.head(5)
```

```python
gold_Compras = silverCompras

def export_dados_excel(resultado):

    pasta = Path("..\\Raw\\table_gold\\")
    pasta.mkdir(parents=True, exist_ok=True)

    if resultado:

        arquivo_excel = pasta / "gold_Compras.xlsx"

        # 1. Criar os arquivos
        gold_Compras.to_excel(arquivo_excel, index=False)


        # 3. Consolidar em um Excel com duas abas
        with pd.ExcelWriter(arquivo_excel, engine="openpyxl") as writer:

            gold_Compras.to_excel(
                writer,
                sheet_name="g_compras",
                index=False
            )

            gold_fornecedor.to_excel(
                writer,
                sheet_name="fornecedor",
                index=False
            )


            gold_cod_qtd.to_excel(
                writer,
                sheet_name="cod_prod",
                index=False
            )

            gold_resp.to_excel(
                writer,
                sheet_name="resposavel",
                index=False
            )

            gold_mes.to_excel(
                writer,
                sheet_name="mes",
                index=False
            )

            gold_ano.to_excel(
                writer,
                sheet_name="ano",
                index=False
            )

            gold_nf.to_excel(
                writer,
                sheet_name="nf",
                index=False
            )

            
        return True

    print("ERRO")
    return False
```

```python
export_dados_excel(True)
```
