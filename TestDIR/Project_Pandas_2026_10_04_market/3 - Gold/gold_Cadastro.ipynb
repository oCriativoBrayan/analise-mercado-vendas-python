## Cria tabelas com dados agrupados

```python
gold_Cadastro = silverCadadastro # criar uma nova tabela com dados importados
```

```python
#cria uma tabelas com dados categoria x vendas
gold_s_venda = (
    gold_Cadastro
    .groupby('CATEGORIA_BRUTA', as_index=False)['VENDA_CADASTRO']
    .sum()
)
```

```python
#cria uma tabelas com dados categoria x custo
gold_s_custo = (
    gold_Cadastro
    .groupby('CATEGORIA_BRUTA', as_index=False)['CUSTO_VENDA']
    .sum()
)
```

```python
#cria uma tabelas com dados categoria x margem
gold_s_margem = (
    gold_Cadastro
    .groupby('CATEGORIA_BRUTA', as_index=False)['MARGEM']
    .sum()
)
```

## Cria 1 tabelas com dados agrupados das 3 anteriores

```python
# Cria uma nova tabela com dados 3 sumatorias, horizontal
CategoriaVendas = pd.concat(
    [gold_s_venda, gold_s_custo, gold_s_margem],
    axis=1
)
```

```python
# Cria uma nova tabela com dados 3 sumatorias, contatenando na vertical
CategoriaVendas_2 = pd.concat(
    [gold_s_venda, gold_s_custo, gold_s_margem],
    ignore_index=True
)
```

```python
# Cria uma nova tabela com dados 3 sumatorias, deixando tabea 1 inteira, apenas resultado das outras 2
CategoriaVendas_3 = pd.concat(
    [gold_s_venda['CATEGORIA_BRUTA'], gold_s_venda['VENDA_CADASTRO'], gold_s_custo['CUSTO_VENDA'], gold_s_margem['MARGEM']],
    axis=1
)
```
