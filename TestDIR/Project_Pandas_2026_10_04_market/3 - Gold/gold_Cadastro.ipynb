```python
def export_dados_excel(resultado):

    pasta = Path("..\\Raw\\table_gold\\")
    pasta.mkdir(parents=True, exist_ok=True)

    if resultado:

        arquivo_excel = pasta / "gold_Cadastro.xlsx"

        # 1. Criar os arquivos
        gold_Cadastro.to_excel(arquivo_excel, index=False)


        # 3. Consolidar em um Excel com duas abas
        with pd.ExcelWriter(arquivo_excel, engine="openpyxl") as writer:

            gold_Cadastro.to_excel(
                writer,
                sheet_name="g_cadastro",
                index=False
            )


            resumo.to_excel(
                writer,
                sheet_name="resumo",
                index=False
            )


        return True

    print("ERRO")
    return False
```

```python
export_dados_excel(True)
```
