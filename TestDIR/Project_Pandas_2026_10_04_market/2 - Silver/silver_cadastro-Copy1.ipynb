```python
def clean_numeric_column(series: pd.Series) -> pd.Series:
    return pd.to_numeric(
        series.astype(str)
              .str.strip()
              .str.replace(',', '.', regex=False),
        errors='coerce'
    ).fillna(0.0)


silverCadastro['VENDA_CADASTRO'] = clean_numeric_column(vt_VendasCadastro)

silverCadastro['CUSTO_VENDA'] = clean_numeric_column(vt_CustoCadastro)
```

```python
def exportar_dados_excel(resultado):

    pasta = Path(r"C:\Users\Public\Teste_Jupyter\Mercado\2 - Silver")
    pasta.mkdir(parents=True, exist_ok=True)

    if resultado:

        arquivo_excel = pasta / "SilverCadastro_ETL.xlsx"
        arquivo_csv = pasta / "SilverCadastro_ETL.csv"

        df_csv = pd.read_csv(arquivo_csv)

        with pd.ExcelWriter(arquivo_excel, engine="openpyxl") as writer:

            SilverCadastro_ETL.to_excel(
                writer,
                sheet_name="xlsx",
                index=False
            )

            df_csv.to_excel(
                writer,
                sheet_name="csv",
                index=False
            )

        return True

    print("ERRO")
    return False
```
