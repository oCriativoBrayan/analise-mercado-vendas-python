```python
REGRAS_CATEGORIA = {
    "FRIOS & LATICINIOS": "FRIOS E LATICINIOS",
    "FRIOS": "FRIOS E LATICINIOS",
    "LATICINIOS": "FRIOS E LATICINIOS",
    "SECOS E MOLHADOS": "SECOS E MOLHADOS",
    "SECOS": "SECOS E MOLHADOS",
    "MOLHADOS": "SECOS E MOLHADOS",
    "HORTI FRUTI": "HORTIFRUTI",
    "HORT": "HORTIFRUTI",
    "HORTI": "HORTIFRUTI",
    "CARNES": "ACOUGUE",
    "CARNES & AÇOUGUE": "ACOUGUE",
    "AÇOUGUE": "ACOUGUE",

}

silverCadastro["CATEGORIA_BRUTA"] = (
    silverCadastro["CATEGORIA_BRUTA"]
    .replace(REGRAS_CATEGORIA)
)
```

```python
def clean_numeric_column(series: pd.Series) -> pd.Series:
    return pd.to_numeric(
        series.astype(str)
              .str.strip()
              .str.replace(',', '.', regex=False),
        errors='coerce'
    ).fillna(0.0)


silverCadastro['VENDA_CADASTRO'] = clean_numeric_column(silverCadastro['VALOR_VENDA_CADASTRO'])

silverCadastro['CUSTO_VENDA'] = clean_numeric_column(silverCadastro['CUSTO_CADASTRO'])
```

```python
silverCadastro = silverCadastro.drop(columns=['VALOR_VENDA_CADASTRO','CUSTO_CADASTRO'])
```
