### Ajusta Data

```python
# Cria coluna Data
DF_Vendas['DATA'] = pd.to_datetime(DF_Vendas['DATA_HORA_VENDA'], format="mixed")

# cria colunas Mes e Ano
DF_Vendas['MES'] = DF_Vendas['DATA'].dt.month
DF_Vendas['ANO'] = DF_Vendas['DATA'].dt.year

# deleta coluna DATA_HORA_VENDA
DF_Vendas = DF_Vendas.drop(columns={'DATA'})

DF_Vendas.head(5)
```

### Validacao Valores de venda total

```python
DF_Vendas['VENDAS_TOTAL'] = (DF_Vendas['QTD_VENDIDA'] * DF_Vendas['PRECO_UNIT_COBRADO']).round(2)
DF_Vendas['DIF_VENDA'] = (DF_Vendas['TOTAL_PAGO'] - DF_Vendas['VENDAS_TOTAL']).round(2)
```

```python
check = (DF_Vendas['QTD_VENDIDA'] * DF_Vendas['PRECO_UNIT_COBRADO'] - DF_Vendas['TOTAL_PAGO']).round(2)
print("fórmula confere:", (check == DF_Vendas['DIF_VENDA'].round(2)).all())
```

```python
zerados = DF_Vendas['TOTAL_PAGO'] == 0

print(DF_Vendas.loc[zerados, 'STATUS_VENDA'].value_counts())
print("DIF = -VENDAS_TOTAL onde TOTAL_PAGO é 0:",
      np.isclose(DF_Vendas.loc[zerados, 'DIF_VENDA'],
                 -DF_Vendas.loc[zerados, 'VENDAS_TOTAL']).all())
```

```python
check = (DF_Vendas['TOTAL_PAGO'] - DF_Vendas['QTD_VENDIDA'] * DF_Vendas['PRECO_UNIT_COBRADO'])
diff = (check - DF_Vendas['DIF_VENDA']).abs()

print("fórmula confere:", (diff < 0.011).all())
print("maior diferença:", diff.max())
```

```python
conc = DF_Vendas[DF_Vendas['STATUS_VENDA'] == 'CONCLUIDA']
print("CONCLUIDA com DIF_VENDA > R$ 0,01:", (conc['DIF_VENDA'].abs() > 0.01).sum())
```

```python
DF_Concluidas = DF_Vendas[DF_Vendas['STATUS_VENDA'] == 'CONCLUIDA'].copy()

print("linhas concluídas:", len(DF_Concluidas))
print("faturamento:", round(DF_Concluidas['TOTAL_PAGO'].sum(), 2))
```

```python
def export_dados_excel(resultado):

    pasta = Path("..\\Raw\\table_silver\\")
    pasta.mkdir(parents=True, exist_ok=True)

    if resultado:

        arquivo_excel = pasta / "silverVendas_ETL.xlsx"

        # 1. Criar os arquivos
        silverVendas_ETL.to_excel(arquivo_excel, index=False)


        # 3. Consolidar em um Excel com duas abas
        with pd.ExcelWriter(arquivo_excel, engine="openpyxl") as writer:

            silverVendas_ETL.to_excel(
                writer,
                sheet_name="Excel",
                index=False
            )


        return True

    print("ERRO")
    return False
```
