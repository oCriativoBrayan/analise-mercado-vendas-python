```python
def export_dados_excel(resultado):

    pasta = Path(r"C:\Users\Public\Teste_Jupyter\Mercado\2 - Silver")
    pasta.mkdir(parents=True, exist_ok=True)

    if resultado:

        arquivo_excel = pasta / "silverVendas_ETL.xlsx"
        arquivo_csv = pasta / "silverVendas_ETL.csv"

        # 1. Criar os arquivos
        silverVendas_ETL.to_csv(arquivo_csv, index=False)
        silverVendas_ETL.to_excel(arquivo_excel, index=False)

        # 2. Ler o CSV criado
        df_csv = pd.read_csv(arquivo_csv)

        # 3. Consolidar em um Excel com duas abas
        with pd.ExcelWriter(arquivo_excel, engine="openpyxl") as writer:

            silverVendas_ETL.to_excel(
                writer,
                sheet_name="Excel",
                index=False
            )

            df_csv.to_excel(
                writer,
                sheet_name="CSV",
                index=False
            )

        return True

    print("ERRO")
    return False
```

```python
export_dados_excel(resultado)
```
